```python
print(pd.crosstab(df['Industry'], df['Layoff_Risk']))

# gyakorisági eloszlás vizsgálata

# szektorok és a leépítési kockázat közötti számszerű kapcsolat
# megmutatja, hogy az adott szektorban pontosan hány munkavállaló esik az egyes kockázati kategóriába
# megmutatja, hogy melyik szektor a legkritikusabb illetve legbiztonságosabb
```

```python
risk_by_industry = pd.crosstab(df['Industry'], df['Layoff_Risk'], normalize='index') * 100
print(risk_by_industry.round(1))

# százalékos arányban mutatja meg
```

```python

crosstab_pct = pd.crosstab(df['Industry'], df['Layoff_Risk'], normalize='index') * 100

ordered_columns = ['Low', 'Medium', 'High']
crosstab_pct = crosstab_pct.reindex(columns=ordered_columns)

crosstab_pct.plot(kind='bar', stacked=True, color=['green', 'orange', 'red'], figsize=(10, 6))

plt.title('Leépítési kockázat százalékos megoszlása szektoronként', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Szektorok (Industry)', fontsize=12)
plt.ylabel('Munkavállalók aránya (%)', fontsize=12)
plt.legend(title='Kockázati szint (Layoff Risk)', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

A legnagyobb kockázat a Manufacturing szektorban van.
Ezt tovább vizsgálva, hogy mi okozhatja, megnézzük, hogy a munkakörök és a automatizáltsági szintet

```python
#Automatizáltság és Kockázat kapcsolata

# új oszlop létrehozása

df['Automation_Bins'] = pd.cut(df['Tasks_Automated_Percentage'],
                               bins=[0, 25, 50, 75, 100],
                               labels=['0-25%', '26-50%', '51-75%', '76-100%'],
                               include_lowest=True)
```

```python
print("\n--- Automatizáltság vs Kockázat (Százalékos megoszlás) ---")
automation_crosstab_pct = pd.crosstab(df['Automation_Bins'], df['Layoff_Risk'], normalize='index') * 100
print(automation_crosstab_pct.round(1))
```

A leépítési kockázatot nem a gazdasági válság vagy a szektorok általános helyzete mozgatja, hanem maga a technológiai kiválthatóság. Nem az a kérdés, hogy ki melyik szektorban dolgozik, hanem az, hogy a napi munkájának mekkora részét képes elvégezni egy algoritmus.

```python
munkakor_kockazat_db = pd.crosstab(df['Job_Role'], df['Layoff_Risk'])

# Oszlopok sorrendjének logikus beállítása (Alacsony -> Közepes -> Magas)
munkakor_kockazat_db = munkakor_kockazat_db[['Low', 'Medium', 'High']]

print("--- Munkakörök és Kockázatok megoszlása (Darabszám) ---")
print(munkakor_kockazat_db)
```

```python
munkakor_kockazat_pct = pd.crosstab(df['Job_Role'], df['Layoff_Risk'], normalize='index') * 100

# Oszlopok sorrendjének beállítása (Low -> Medium -> High)
munkakor_kockazat_pct = munkakor_kockazat_pct[['Low', 'Medium', 'High']]

print("--- Munkakörök és Kockázatok százalékos megoszlása (%) ---")
print(munkakor_kockazat_pct.round(1)) # 1 tizedesjegyre kerekítve a tisztaságért
```

```python
# kétváltozós feltáró elemzés

data = {
    'Job_Role': [
        'Support Specialist', 'Operator', 'Store Manager', 'ML Engineer', 'Production Supervisor',
        'Financial Analyst', 'Auditor', 'Sales Associate', 'Software Engineer', 'Inventory Analyst',
        'Academic Coordinator', 'Network Engineer', 'Dispatcher', 'Accountant', 'Health Analyst',
        'Research Assistant'
    ],
    'Avg_Automation': [5.0, 7.0, 16.5, 20.0, 27.3, 20.0, 42.7, 44.0, 45.0, 67.0, 67.0, 45.3, 50.5, 60.5, 68.0, 69.0],
    'Main_Risk': [
        'Low', 'Low', 'Low', 'Low', 'Low',
        'Medium', 'Medium', 'Medium', 'Medium', 'Medium', 'Medium',
        'High', 'High', 'High', 'High', 'High'
    ]
}

df_plot = pd.DataFrame(data).sort_values(by='Avg_Automation')

plt.figure(figsize=(12, 8))


color_map = {'Low': 'green', 'Medium': 'orange', 'High': 'red'}


colors = [color_map[risk] for risk in df_plot['Main_Risk']]


bars = plt.barh(df_plot['Job_Role'], df_plot['Avg_Automation'], color=colors, height=0.6)


for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 1,
        bar.get_y() + bar.get_height()/2,
        f'{width:.1f}%',
        va='center',
        ha='left',
        fontsize=10,
        fontweight='bold',
        color='black'
    )


plt.title('Munkakörök átlagos automatizáltsága és leépítési kockázata', fontsize=14, fontweight='bold', pad=20)
plt.xlabel('Átlagos automatizált feladatok aránya (%)', fontsize=11, fontweight='bold')
plt.ylabel('Munkakörök (Job Role)', fontsize=11, fontweight='bold')
plt.xlim(0, 80)
plt.grid(axis='x', linestyle='--', alpha=0.5)


from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='green', label='Alacsony kockázat (Low)'),
    Patch(facecolor='orange', label='Közepes kockázat (Medium)'),
    Patch(facecolor='red', label='Magas kockázat (High)')
]
plt.legend(handles=legend_elements, title='Leépítési kockázat szintje', loc='lower right', frameon=True)

plt.tight_layout()
plt.show()
```

A gragikon feladatok automatizáltságának mértéke és a leépítési kockázat közötti összefüggést mutatja meg. Munkakörök sorbarendezve a szerint, hogy melyik a legkevésbé vagy leginkább automatizálható. A másik tengelyen pedig azt mutatja, hogy az adott munkakörben a feldatok hány százaléka váltható ki AI-al. A színek a leépítés kockázatát mutatják. A piros és a narancs zóna a legveszélyeztetettebb területek. De az is látható, hogy önmagában nem csak az automatizáltság az egyetlen befolyásoló tényező ami a leépítést meghatározza.
